```python
from graphviz import Digraph
```

```python
### SIR DAG

sir_dag = Digraph('dag', graph_attr={'rankdir': 'LR'})

for compartment in ['S', 'I', 'R']:
    sir_dag.node(compartment, compartment)

sir_dag.edge('S', 'I', 'βS')

sir_dag.edge('I', 'R', 'γI')

sir_dag.edge('R', 'S', 'ωR')


sir_dag.render('./figures/assignment2/SIRS_dag', format = 'jpg')
```
